```python
import joblib
import pandas as pd

model = joblib.load("models/random_forest.pkl")
scaler = joblib.load("models/scaler.pkl")
feature_columns = joblib.load("models/feature_columns.pkl")
df = pd.read_csv("data/feature_engineered_data.csv")
```

```python
sample = df.iloc[[0]].copy()
sample = sample.drop(columns=["loan_status"])
```

```python
ratios = []

for ratio in [0.05,0.10,0.15,0.20,0.25,0.30,0.35,0.40,0.45,0.50,0.55,0.60,0.65,0.70]:

    temp = sample.copy()

    temp["loan_percent_income"] = ratio

    probability = model.predict_proba(temp)[0][1]

    prediction = model.predict(temp)[0]

    ratios.append({
        "Ratio": ratio,
        "Probability": probability,
        "Prediction": prediction
    })

pd.DataFrame(ratios)
```
